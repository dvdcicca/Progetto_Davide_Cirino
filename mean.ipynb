In questo script gestisco il calcolo dell'energia media e della varianza

Plottiamo anche il grafico dell'energia con $\alpha_{1}$ fissato a 2.5

```python
t = []
E = []
with open("vmc.csv") as f:
    ff = csv.reader(f)
    title = next(ff)
    
    for row in ff:
        floats = [float(i) for i in row]
        t.append(floats[0])
        E.append(floats[1])

plt.plot(t, E)
    
    
```
